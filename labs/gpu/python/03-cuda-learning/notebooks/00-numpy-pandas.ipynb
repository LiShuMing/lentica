# NumPy and pandas: CPU basics

Start here while the CUDA packages are installing. These exercises use CPU memory.

```
import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print(sys.executable)
print("NumPy", np.__version__, "| pandas", pd.__version__)
```

## Arrays, shapes, slicing, and broadcasting

```
a = np.arange(12, dtype=np.float32).reshape(3, 4)
print("Array:\n", a)
print("Shape:", a.shape, "| dtype:", a.dtype)
print("Second column:", a[:, 1])
print("Broadcast:\n", a + np.array([10, 20, 30, 40]))
print("Row means:", a.mean(axis=1))
```

## Filter and aggregate a DataFrame

```
sales = pd.DataFrame(
    {
        "region": ["East", "West", "East", "North", "West"],
        "quantity": [2, 5, 3, 4, 2],
        "price": [10.0, 8.0, 15.0, 9.0, 12.0],
    }
)
sales["revenue"] = sales["quantity"] * sales["price"]
print(sales)
print("\nRevenue >= 30:\n", sales.loc[sales["revenue"] >= 30])
summary = sales.groupby("region", as_index=False)["revenue"].sum()
print("\nBy region:\n", summary)
```

## Plot the result

```
fig, ax = plt.subplots(figsize=(6, 3))
ax.bar(summary["region"], summary["revenue"], color=["#2c8c73", "#bf4664", "#4d78b7"])
ax.set(xlabel="Region", ylabel="Revenue", title="Revenue by region")
fig.tight_layout()
plt.show()
```
