```
x = np.random.randn(10000, 128)
p = np.random.randn(128)

import time

# loop version
start = time.time()
dist = np.zeros(x.shape[0])
for i in range(x.shape[0]):
    dist[i] = np.sqrt(np.sum((x[i] - p) ** 2))
end = time.time()
print("Time taken for distance calculation:", end - start, "seconds")
# simd version
start = time.time()
dist = np.sqrt(np.sum((x - p) ** 2, axis=1))
end = time.time()
print("Time taken for distance calculation:", end - start, "seconds")

import numexpr as ne

start = time.time()
# 只需要把表达式写成字符串，速度能发生质的飞跃
dist = np.sqrt(ne.evaluate("sum((x - p)**2, axis=1)"))
end = time.time()
print("Time taken for distance calculation:", end - start, "seconds")
```
